## CLAHE
CLAHE es una técnica de mejora de contraste que:
- Divide la imagen en pequeñas regiones (tiles)
- Aplica ecualización de histograma a cada región
- Limita el contraste para evitar sobre-amplificación
- Es especialmente útil para imágenes médicas y de baja calidad
